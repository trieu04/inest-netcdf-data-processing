```python
#
#print(ds)
sample = ds.isel(time=0, lat=0, lon=0)
print(sample)

```

```python
# Define lat/lon range
lat_min, lat_max = 15, 20
lon_min, lon_max = 99, 120

# Subset the dataset
subset = ds.sel(lat=slice(lat_min, lat_max), lon=slice(lon_min, lon_max))

# Initialize dictionary
output = {}

# Loop over lat-lon pairs
for lat in subset.lat.values:
    for lon in subset.lon.values:
        # Select by value for lat/lon and index for time
        sample = subset.sel(lat=lat, lon=lon).isel(time=0)

        # Extract data variables
        data_dict = {var: float(sample[var].values) for var in sample.data_vars}

        key = f"{lat:.2f},{lon:.2f}"
        output[key] = data_dict

# Convert to JSON
json_output = json.dumps(output, indent=2)

# Save (optional)
with open("grid_data.json", "w") as f:
    f.write(json_output)
```
